```python
# 1. Person → Student, Teacher
# Question
# Create a parent class Person with attributes name and age. Create two child classes Student
# and Teacher that inherit from Person.
# Concept
# • Hierarchical Inheritance
# • Parent attributes inherited by multiple child classes.
# How to Solve
# 1. Create a parent class Person.
# 2. Add name and age using a constructor.
# 3. Create Student(Person).
# 4. Create Teacher(Person).
# 5. Create objects of both classes.
# 6. Show that both children can access name and age.
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Student(Person):
    def __init__(self,name,age):
        Person.__init__(self,name,age)
    def display(self):
        print("Name= ",self.name)
        print("Age  ",self.age)
class Teacher(Person):
    def __init__(self,name,age):
        Person.__init__(self,name,age)
    def display_tec(self):
        print("Name= ",self.name)
        print("Age  ",self.age)

std=Student("Vaishnavi",56)
std.display()

tec=Teacher("Mangesh",89)
tec.display_tec()
    
```

```python
# 15. Vehicle Constructor
# Concept
# Constructor inheritance.
# How to Solve
# 1. Parent stores brand.
# 2. Car adds doors.
# 3. Bike adds engine_cc.
# 4. Use super().

class Vehicle:
    def __init__(self,brand):
        self.brand=brand
class Car(Vehicle):
    def __init__(self,brand,doors):
        Vehicle.__init__(self,brand)
        self.doors=doors
    def display(self):
        print("The brand= ",self.brand)
        print("Doors=    ",self.doors)
class Bike(Vehicle):
    def __init__(self,brand,engine_cc):
        Vehicle.__init__(self,brand)
        self.engine_cc=engine_cc
    def display(self):
        print("The brand= ",self.brand)
        print("Engine_cc= ",self.engine_cc)
car=Car("AUDI","550cc")
car.display()
bike=Bike("BMW","750cc")
bike.display()
```

```python
# 16. Shape Area
# Concept
# Polymorphism using overriding.
# How to Solve
# 1. Parent constructor.
# 2. Circle calculates πr2.
# 3. Square calculates side2.
# 4. Override area().
class Shape_area:
    def area(self):
        print("The shapes area calculator")
class Circle(Shape_area):
    def __init__(self,r):
        self.r=r
    def area(self):
        area=2*3.14*self.r**2
        print("The Circle area is = ",area) 
class Square(Shape_area):
    def __init__(self,side):
        self.side=side
    def area(self):
        area=self.side**2
        print("the square area is = ",area)
cir=Circle(5)
cir.area()
squ=Square(5)
squ.area()
```

```python
# 24. Bank System
# Concept
# Three children inherit one parent.
# How to Solve
# 1. Parent → Account
# 2. Store balance.
# 3. Savings → interest.
# 4. Current → overdraft.
# 5. FixedDeposit → maturity.
# 6. Show each separately.
class Account:
    def __init__(self,balance):
        self.balance=balance
class Saving(Account):
    def __init__(self,balance):
        Account.__init__(self,balance)
        
    def display(self):
        intrest=self.balance*0.05
        print("The Balance is= " ,self.balance)
        print("Intrest =    ",intrest)
class Current(Account):
    def __init__(self,balance):
        Account.__init__(self,balance)
        
    def display(self):
        overdraft=5000
        print("balance       =    ",self.balance)
        print("The overdraft limit = " ,overdraft)
       
class FixedDeposit(Account):
    def __init__(self,balance):
        Account.__init__(self,balance)
        
    def display(self):
        maturity=self.balance*1.08
        print("Balance       =    ",self.balance)
        print("The maturity is= " ,maturity)
        
save=Saving(3000)
save.display()
curr=Current(300)
curr.display()
fixed=FixedDeposit(300)
fixed.display()

```

```python
# 27. Employee Salary
# Concept
# Different salary calculations.
# How to Solve
# 1. Parent stores common data.
# 2. FullTime calculates monthly salary.
# 3. PartTime calculates hourly salary.
# 4. Override calculation method.

class Employee:
    def __init__(self,name):
        self.name=name
class FullTime(Employee):
    def __init__(self,name,Mon_sal):
        Employee.__init__(self,name)
        self.Mon_sal=Mon_sal
    def Monthly(self):
        print("The monthly salary is = ",self.Mon_sal)
class PartTime(Employee):
    def __init__(self,name,hours,rate):
        Employee.__init__(self,name)
        self.hours=hours
        self.rate=rate
        
    def Hourly(self):
        salary=self.hours*self.rate
        print("The hourly salary is = ",salary)

full=FullTime("vani",23000)
full.Monthly()
print()
part=PartTime("vidhi",7,5000)
part.Hourly()
print()
```
