```python
# 12. Class Name: Account → LoanAccount
# Problem
# Create a parent class named Account.
# Explanation
# Store:
# • balance
# Create subclass:
# • LoanAccount
# Add:
# • calculate_emi()
class Account:
    def __init__(self,balance):
        self.balance=balance
        
class LoanAccount(Account):
    def __init__(self,balance):
        super().__init__(balance)
        
    def calculate_emi(self,anual_rate,months):
        r=anual_rate/(12*100)
        emi=(self.balance*r*(1+r)**months)/((1+r)**months-1)
        print("Loan amt= ",self.balance)
        print("Monthly EMI= ".format(emi))

loan = LoanAccount(500000)
loan.calculate_emi(10, 60)
        
```

```python
# 16. Class Name:
# Person → Employee → Manager
# Problem
# Create a three-level inheritance hierarchy.
# Explanation
# Person
# Store:
# • name
# ↓
# Employee
# Add:
# • employee_id
# ↓
# Manager
# Add:
# • department
# Display all information.
class Person:
    def __init__(self,name):
        self.name=name
class Employee(Person):
    def __init__(self,name,emp_id):
        self.emp_id=emp_id
        Person.__init__(self,name)
class Manager(Employee):
    def __init__(self,name,emp_id,dept):
        self.department=dept
        Employee.__init__(self,name,emp_id)
    def show_all(self):
        print("Name =       ",self.name)
        print("emp_id =     ",self.emp_id)
        print("department = ",self.department)
man=Manager("vaishnavi",34,"data science")
man.show_all()
    
```

```python
# 26. Class Name:
# Account → SavingAccount → SeniorCitizenAccount
# Problem
# Create multi-level inheritance.
# Explanation
# Each level should implement different interest rules.
class Account:
    def __init__(self, amount):
        self.amount = amount
    def A_interst(self):
        return self.amount*0.15
class SavingAccount(Account):
    def __init__(self, amount):
        self.amount = amount
    def S_interst(self):
        return self.amount*0.10
class SeniorCitizenAccount(SavingAccount):
    def __init__(self,amount):
        self.amount=amount
    def interst(self):
        return self.amount*0.5
    def show(self):
        print("Intrest on Account             = ",self.A_interst())
        print("Intrest on SavingAccount       = ",self.S_interst())
        print("Intrest on SeniorCitizenAccount= ",self.interst())
sen=SeniorCitizenAccount(3000)
sen.show()
```

```python
# 29. Class Name:
# Shape → TwoDShape → Triangle
# Problem
# Create multi-level inheritance.
# Explanation
# Triangle should calculate its area using Heron's Formula.
# Store:
# • side1
# • side2
# • side3
# Implement the area calculation.

class Shape:
    def show(self):
        print("The shape is Triangle")
class TwoDShape(Shape):
    def shape(self):
        print("2 D shape = Triangle")
class Triangle(TwoDShape):
    def __init__(self,side1,side2,side3):
        self.side1=side1
        self.side2=side2
        self.side3=side3
    def area(self):
        area=(self.side1+self.side2+self.side3)/2
        print("The area of triangle is = ",area)

tri=Triangle(5,6,7)
tri.area()
```

```python
# 30. Class Name:
# GrandParent → Parent → Child → GrandChild
# Problem
# Create four levels of inheritance.
# Explanation
# Each class should have its own method.
# The GrandChild object should be able to access methods from all parent classes.
# Demonstrate Python's Method Resolution Order (MRO) using:
# ClassName.mro()
# or
# help(ClassName)
class GrandParent:
    def eat(self):
        print("Grand parent eats")
class Parent(GrandParent):
    def walk(self):
        print("Parents walk")
class Child(Parent):
    def ride(self):
        print("Child ride bike")     
class GrandChild(Child):
    def crawl(self):
        print("Grandchild crawls")

inhe=GrandChild()
inhe.eat()
inhe.walk()
inhe.ride()
inhe.crawl()

print(GrandChild.__mro__)
```
