```python
# 15.Create a complete business-style visualization using Matplotlib:
# o X-axis → Month
# o Y-axis → Sales
# o Plot monthly sales
# o Add title
# o Add X/Y labels
X=["jan","feb","march","april","may","june","july","aug","sept","oct","nov","dec"]
Y=[12,23,45,6,789,89,780,98,6,54,24,56]
plt.plot(X,Y,
        color="purple",
        linewidth=1,
        marker="^",
        markersize=10,
        markeredgecolor="yellow",
        markerfacecolor="red")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid(True,alpha=0.5)
plt.legend(["Sales"])
plt.figure(figsize=(10,6),
          dpi=67,
          facecolor="pink",
          edgecolor="yellow")
plt.show()
```

```python
# 3. Create a bar chart showing the revenue of 5 different products. Display the
# revenue value on top of each bar.
prod=["pen","pencil","book","pen case","Bottel"]
rev=[200,300,400,40,500]
plt.bar(prod,rev)
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.title("Producs Revenue")
for i in range(len(prod)):
    plt.text(i,rev[i],rev[i],ha="center")
plt.show()
```

```python
# 9. Plot three mathematical functions on the same graph:
# Add different colors, markers, line styles, and a legend.
x=np.arange(1,6)

plt.plot(x,x**2,color="red",linestyle="-",marker="o",label="y=x")
plt.plot(x,x**3,color="yellow",linestyle="--",marker="*",label="y=x")
plt.plot(x,x**4,color="purple",linestyle="-",marker="^",label="y=x")
plt.show()
```

```python
# 10.Create a 2×2 subplot containing:
# o Line chart
# o Bar chart
# o Scatter plot
# o Histogram
x=np.linspace(1,10,100)
fig,axes=plt.subplots(2,2,
           figsize=(10,20))

axes[0,0].plot(x,x**2,color="blue")
axes[0,0].set_title("Square plot")

axes[0,1].bar(x,np.sqrt(x),color="purple")
axes[0,1].set_title("Square root plot")

axes[1,0].scatter(x,np.tan(x),color="orange")
axes[1,0].set_title("Tan plot")

axes[1,1].hist(x,np.log10(x),color="black",linewidth=6)
axes[1,1].set_title("Log plot")

plt.show()
```

```python
# 11.Create a monthly sales chart and identify the maximum-sales month.
# Highlight that point using annotate().
months=["jan","feb","march","april","may","june","july","aug","sept","oct","nov","dec"]
sales=[10,10,20,20,45,6,7,8,9,3,20,45]
plt.plot(months,sales,
        marker="^")
m=max(sales)
i=sales.index(m)

plt.annotate("Max",(months[i],m))

plt.show()
```

```python
# 12.Create a scatter plot of Age vs Income. Highlight customers whose income is
# greater than ₹1,00,000 using a different marker.
age=[23,45,6,7,8,89]
income=[2000,3000,65222432,34567,9876,1112345]

for i in range(len(age)):
    if income[i]>100000:
        plt.scatter(age[i],income[i],
                    color="orange",
                    marker="^",
                   s=100)
    else:
        plt.scatter(age[i],income[i],
                   color="blue",
                   marker="o",)

plt.xlabel("Age")
plt.ylabel("Income")
        
plt.scatter(age,income)
```

```python
# 13.Create a stacked bar chart showing monthly sales from:
# o Online
# o Offline
# o Corporate
online=[100,100,1000,1000]
offline=[200,300,100,1000]
corporate=[300,200,3000,20]
months=["jan","feb","march","april"]

plt.bar(months,online,label="Online")
plt.bar(months,offline,bottom=online,label="Offline")
plt.bar(months,corporate,bottom=[online[i]+offline[i] for i in range(len(online))],label="corporate")
plt.legend()
plt.show()
```

```python

# 14.Create a graph containing actual sales and target sales for 12 months.
# Highlight the months where actual sales are below the target.
actual=[12,23,45,6,7,8,9,0,4,5,6,7]
target=[1,2,3,4,5,6,7,8,9,0,1,2]
months=["jan","feb","march","april","may","june","july","aug","sept","oct","nov","dec"]
plt.plot(months,actual,label="Actual")
plt.plot(months,target,label="Target")

for i in range(len(months)):
    if actual[i]<target[i]:
        plt.scatter(months[i],actual[i],color="red")
plt.title("Actual VS Target Sales")
plt.show()
```

```python
# 15.Create a mini analytics dashboard using Matplotlib with 4 subplots:
# o Monthly revenue line chart
# o Category-wise sales bar chart
# o Customer-age histogram
# o Revenue vs discount scatter plot

months=["jan","feb","march","april","may","june","july","aug","sept","oct","nov","dec"]
revenue=[1000,200000,30000,981276,231445,123456,233456789,987654,100000,500172,921111,456000]

categ=["online","offline","coorporate","shop"]
sales=[56,78,90,10]

cust=["vidnesh","mangesh","shreya","palak","sakhi"]
age=[12,45,67,87,65]

rev=[2000,3000,4000,5000,6000]
disc=[10,20,30,40,50]

fig,axes=plt.subplots(2,2,figsize=(10,20))

axes[0,0].plot(months,revenue,"pink",linewidth=4)
axes[0,0].set_title("Month Revenue plot")

axes[0,1].plot(categ,sales,"yellow",linewidth=4)
axes[0,1].set_title("Category sales plot")

axes[1,0].plot(cust,age,"green",linewidth=4)
axes[1,0].set_title("Customer age plot")

axes[1,1].plot(rev,disc,"orange",linewidth=4)
axes[1,1].set_title("Revenue discount plot")
```
