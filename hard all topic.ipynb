```python
# 2. Create a program using nested loops and break/continue that prints all prime
# numbers between two user-given numbers, and also counts how many primes were
# found.
n1=int(input("Enter 1 number"))
n2=int(input("Enter other number"))
p_count=0
for i in range(n1,n2):
    if i<2:
        continue
    for j in range(2,i):
        if j%1==0:
            break
        else:
            print(j)
            p_count+=1
print("Total prime numbers= ",p_count)
```

```python
# 5. Create abstract class Shape with abstract methods area() and perimeter().
# Implement Circle and Rectangle using super(). Add a class attribute to count total
# shapes created.
from abc import ABC, abstractmethod
class Shape(ABC):
    count=0
    def __init__(self):
        Shape.count+=1
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass
class Circle(Shape):
    def __init__(self,r):
        super().__init__()
        self.r=r
    def area(self):
        return 3.14*self.r**2
    def perimeter(self):
        return 2*3.14*self.r
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width 
    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)
c=Circle(5)
rect=Rectangle(3,5)
print("Circle area= ",c.area())
print("Circle perimeter = ",c.perimeter())
print("REctangle area= ",rect.area())
print("Retangle perimeter= ",rect.perimeter())    
print("Shape count= ",Shape.count)
```

```python
# 9. Create a class ComplexNumber and overload +, -, *, and == operators completely.
class ComplexNumber:
    def __init__(self,a):
        self.a=a
    def __add__(self,other):
        return self.a+other.a
    def __sub__(self,other):
        return self.a-other.a
    def __mul__(self,other):
        return self.a*other.a
    def __eq__(self,other):
        return self.a==other.a
c1=ComplexNumber(9)
c2=ComplexNumber(6)
print(c1+c2)
print(c1-c2)
print(c1*c2)
print(c1==c2)
```

```python
# 10. Create a class Student with private marks list.
# • Add setter with validation (each mark 0-100)
# • Calculate average
# • Overload > operator to compare two students based on average
class Student:
    def __init__(self,marks):
        self.__marks=marks
    def get_marks(self):
        return self.__marks
    def set_marks(self,new_m):
        for i in new_m:
            if i<0 or i>100:
                print("Invalid")
                return    
        self.__marks=new_m
        print("Marks are= ",self.__marks)
    def avg(self):
        return sum(self.__marks)/len(self.__marks)
        print("The average of marks is= ",avg)
    def __gt__(self,other):
        return self.avg()>other.avg()
std1=Student([56,78,34])
std2=Student([67,90,89])
print(std1>std2)
std1.set_marks([45,67,90])
std2.set_marks([23,100,56])
std1.avg()
std2.avg()
```

```python
# 19. Create single, multi-level and hierarchical inheritance in one program and show
# the difference in method resolution.
class Person:
    def show(self):
        print("I am Person")
class Student(Person):
    def study(self):
        print("Student is studying")
    def show(self):
        print("I am Student")
class Monitor(Student):
    def manage(self):
        print("Monitor manages class")
class Teacher(Person):
    def teach(self):
        print("Teacher is teaching")
    def show(self):
        print("I am Teacher")
s = Student()
m = Monitor()
t = Teacher()
s.show()
s.study()
m.show()
m.study()
m.manage()
t.show()
t.teach()
print(Student.mro())
print(Monitor.mro())
print(Teacher.mro())
```
