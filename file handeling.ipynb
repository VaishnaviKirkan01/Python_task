```python
# 24. Write a program to encrypt the content of a file (simple character shifting) and save it in
# another file.
file=open("Std.txt","r")
data=file.read()
file.close()

encrypt=""
for ch in data:
    encrypt=chr(ord(ch)+1)
file=open("Std.txt","w")
file.write(encrypt)
file.close()
print("File encrypted successfully")
```

```python
# 25. Read a file containing student marks, calculate average, and write the result in another
# file.
file=open("Std.txt","w")
file.write("99,89,99\n")
file.write("45,67,90")
file.close()
file=open("Std.txt","r")
oth_file=open("new_std.txt","w")
marks=[]
for line in file:
    data=line.strip().split(',')
    for i in data:
        marks.append(int(i))

cal_avg=sum(marks)/len(marks)
print("Average= ",cal_avg)
oth_file.write("Average= "+str(cal_avg))
file.close()
oth_file.close()
```

```python
# 28. Read a file and write only the unique lines into another file (remove duplicates).
file=open("Tenlines.txt","r")
new_file=open("New_file","w")
unique=set()
for line in file:
    if line not in unique:
        unique.add(line)
        new_file.write(line)
print(file.read())
file.close()
new_file.close()
```

```python
# 29. Create a menu-driven program:
# • Write to file
# • Read file
# • Append to file
# • Exit
while True:
    print("Write your choice")
    print("1. Write ")
    print("2. Read")
    print("3. Append")
    print("4. Exit")
    choice=int(input("enter choice number"))
    if choice==1:
    
        file=open("Menu.txt",'w')
        write=input("write data to input")
        file.write(write)
        file.close()
        print("file write successfull")
    elif choice==2:
        file=open("Menu.txt",'r')
        print(file.read())
        print("File read successfull")
    elif choice==3:   
        file=open("Menu.txt",'a')
        app=input("write text to append")
        file.write("\n"+app)
        file.close()
        print("file append sucssful")
    elif choice==4:
        print("Program exited")
        break
    else:
        print("Invalid choice")
        
```
