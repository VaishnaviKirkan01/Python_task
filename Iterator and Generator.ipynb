```python
# Level 2: Intermediate Iterator Problems
# Problem 6
# Create a custom iterator that prints numbers from 1 to 10.
class Numbers:
    def __init__(self,start):
        self.current=start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current>10:
            raise StopIteration
        value=self.current
        self.current+=1
        return value
for num in Numbers(1):
    print(num)
        
```

```python
# Create a custom iterator that prints only even numbers from 2 to 20.
class EvenNumbers:
    def __init__(self,start):
        self.current=start
    def __iter__(self):
        return self
    def __next__(self):
        while self.current<=20:
            if self.current%2==0:
        
                value=self.current
                self.current+=1
                return value
            self.current+=1
        raise StopIteration
for num in EvenNumbers(2):
    print(num)
    
        
        
```

```python
# Create a custom iterator that prints numbers in reverse order from
# 10 to 1.
class Reverse:
    def __init__(self,start):
        self.current=start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current<=0:
            raise StopIteration
        value=self.current
        self.current-=1
        return value
for num in Reverse(10):
    print(num)
```

```python
# Problem 9
# Create a custom iterator that prints the characters of the word
# "GENERATOR" one by one.
class String:
    def __init__(self,word):
        self.word="GENERATOR"
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=len(self.word):
            raise StopIteration
        value=self.word[self.index]
        self.index+=1
        return value
for i in String(1):
    print(i)
    
        
```

```python
# Problem 10
# Create a custom iterator that prints the multiplication table of 5.
# Expected Output:1,10,15,10,15,30,35,40,45
class Multiplication:
    def __init__(self,start):
        self.current=start
    def __iter__(self):
        return self
    def __next__(self):
        while self.current<=50:
            if self.current%5==0:
                value=self.current
                self.current+=1
                return(value)
            self.current+=1
        raise StopIteration
for i in Multiplication(1):
    print(i)
            
            
```

```python
# Problem 22
# Create a custom iterator that prints all multiples of 3 up to 30.
class Multiples:
    def __init__(self,start):
        self.current=start
    def __iter__(self):
        return self
    def __next__(self):
        while self.current<=30:
            if self.current%3==0:
                value=self.current 
                self.current+=1
                return value
            self.current+=1
        raise StopIteration
       
for i in Multiples(3):
    print(i)
```

```python
# Create a custom iterator that prints all alphabet letters from A toz
class Alphabet():
    def __init__(self,start):
        self.current=ord(start)
    def __iter__(self):
        return self
    def __next__(self):
        if self.current>ord("Z"):
            raise StopIteration
        value=chr(self.current)
        self.current+=1
        return value
for i in Alphabet("A"):
    print(i)
            
```

```python
# Problem 24
# Create a custom iterator that prints all vowels from the string below.
text = "Artificial Intelligence"
class vowels:
    def __init__(self,word):
        self.word=word
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index<len(self.word):
            value=self.word[self.index]
            self.index+=1
            if value.lower()in "aeiou":
                return value
        raise StopIteration
        
for i in vowels(text):
    print(i)
```

```python
# Problem 25
# Create a custom iterator that prints the Fibonacci sequence up to 10
# terms.
class Fibonacci:
    def __init__(self,a,b):
        self.a=a
        self.b=b
        self.count=0
        
    def __iter__(self):
        return self
    def __next__(self):
        if self.count>=10:
            raise StopIteration
           
        value=self.a
        self.c=self.a+self.b
        self.a=self.b
        self.b=self.c
        self.count+=1
        return value
for i in Fibonacci(2,5):
    print(i)
        
        
```

```python
# Level 6: Advanced Generator Problems
# Problem 26
# Create a generator that yields the Fibonacci sequence up to 15
# terms.
def Fibonacci():
    a=0
    b=1
    for i in range(15):
        yield a
        c=a+b
        a=b
        b=c
fibo=Fibonacci()
for i in fibo:
    print(i)
    
```

```python
# Problem 27
# Create a generator that yields all prime numbers between 1 and 100.
def PrimeNum():
    for i in range(2,101):
        prime=True
        for j in range(2,i):
            if i%j==0:
                prime=False
                break
        if prime:
            yield i
for i in PrimeNum():
    print(i)
                
        
        
```

```python
# Problem 28
# Create a generator that yields all factors of a given number.
# Example Input:
# num = 24
def Factors(num):
    for i in range(1,num+1):
        if num%i==0:
            yield i
num=24
for i in Factors(num):
    print(i)
        
        
    
    
```
