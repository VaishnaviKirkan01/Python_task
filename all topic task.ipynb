```python
# 14. Create class BankAccount with private attribute __balance. Add deposit and withdraw
# methods with validation.
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance
    def get_bal(self):
        return f"Balance is= {self.__balance}"
    def set_bal(self,new_bal):
        if new_bal>0:
            self.__balance=new_bal
            print("updated balance= ",self.__balance)
        else:
            print("Invalid")
    def deposit(self,dept):
        if dept>0:
            self.__balance+=dept
            print("Deposited balance is= ",self.__balance)
        else:
            print("Enter valid amount")
    def withdraw(self,withd):
         if withd>0:
            self.__balance-=withd
            print("balance after withdrawn= ",self.__balance)
         else:
            print("Enter valid amount")

bank=BankAccount(1000)
bank.deposit(200)
bank.withdraw(200)
    
```

#### Inheritance + Abstraction + super() (19-24)

```python
# 19. Create single level inheritance: Person → Student. Add attributes in both classes.
class Person:
    def __init__(self,name,**kwargs):
        super().__init__(**kwargs)
        self.name=name
class Student(Person):
    def __init__(self,name,marks):
        super().__init__(name)
        self.marks=marks
    def display(self):
        print("Name= ",self.name)
        print("Marks= ",self.marks)

std=Student("vanshaj",67)
std.display()
```

```python
# 20. Create multi-level inheritance: Person → Employee → Manager.
class Person:
    def __init__(self,name,):
        
        self.name=name
class Employee(Person):
    def __init__(self,name,emp_id,):
        super().__init__(name)
        self.emp_id=emp_id
class Manager(Employee):
    def __init__(self,name,emp_id,salary,):
        super().__init__(name,emp_id)
        self.salary=salary

    def display(self):
        print("Name= ",self.name)
        print("Emp_id= ",self.emp_id)
        print("Salary= ",self.salary)
        
man=Manager("vanshaj",67,567000)
man.display()
```

```python
# 21. Create hierarchical inheritance: Parent Animal → Children Dog and Cat.
class Animal:
    def sound(self):
        print("Animal makes sound")
        
        
class Dog(Animal):
    def sound(self):
        print("Dog barks")
        
class Cat(Animal):
    def sound(self):
        print("Cat meows")
    

   
dog=Dog()      
cat=Cat()
dog.sound()
cat.sound()
```

```python
# 22. Create multiple inheritance with two parent classes Father and Mother, and child class
# Child.
class Father:
    def __init__(self,skill):
        self.skill=skill
class Mother:
    def __init__(self,skil):
        self.skil=skil
class Child(Father,Mother):
    def __init__(self,skill,skil,quality):
        Father.__init__(self,skill)
        Mother.__init__(self,skil)
        self.quality=quality
    def display(self):
        print("Father has skill= ",self.skill)
        print("Mother has skill= ",self.skil)
        print("Child has skill= ",self.quality)

ch=Child("Money","cooking","Smart")
ch.display()
```

```python
# 23. Create abstract class Shape with abstract method area(). Implement it in Circle and
# Rectangle using super().
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def area(self,r):
        area=3.14*r**2
    
        print("Circle area= ",area)
class Rectangle(Shape):
    def area(self,length,width):
        area=length*width
    
        print("Rectangle area= ",area)

cir=Circle()
cir.area(4)
rect=Rectangle()
rect.area(7,8)
        
    
```

```python
# 26. Create parent class Animal with method sound(). Override it in Dog and Cat.
class Animal:
    def sound(self):
        print("Animal makes sound")
        
        
class Dog(Animal):
    def sound(self):
        print("Dog barks")
        
class Cat(Animal):
    def sound(self):
        print("Cat meows")
    

   
dog=Dog()      
cat=Cat()
dog.sound()
cat.sound()
```

```python
# 27. Create class Point and overload + operator to add two points.
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return self.x+other.x,self.y+other.y
p1=Point(3,4)
p2=Point(4,3)
print(p1+p2)
```

```python
# 28. Create class Book and overload > operator to compare number of pages of two books.
class Book:
    def __init__(self,page):
        self.page=page
    def __gt__(self,other):
        return self.page>other.page
pg1=Book(4)
pg2=Book(7)
print(pg1>pg2)
```

```python
# 29. Create two different classes Duck and Person both having method walk(). Demonstrate
# Duck Typing.
class Duck:
    def walk(self):
        print("duck flies and walk")
class Person:
    def walk(self):
        print("Person only walk")

d=Duck()
p=Person()

def step(obj):
    obj.walk()
step(d)
step(p)
```

```python
# 30. Create class ComplexNumber and overload both + and - operators.
class ComplexNumber:
    def __init__(self,a):
        self.a=a
    def __add__(self,other):
        return self.a+other.a
    def __sub__(self,other):
        return self.a-other.a
c1=ComplexNumber(8)
c2=ComplexNumber(4)
print("Addition=     ",c1+c2)
print("Substraction =",c1-c2)
```
