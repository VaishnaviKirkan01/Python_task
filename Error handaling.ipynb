```python
# 21. Create a custom exception InsufficientBalanceError for a banking program.
class InsufficientBalanceError(Exception):
    pass
class Bank :
    def __init__(self,amt):
        self.amt=amt
    def withdraw(self,new):
        if new>0:
            
            print("Withdrawn amount= ",new)
        else:
            print("Invalid")
        try:
            if new>=self.amt:
                raise InsufficientBalanceError("Not enough balance")
            else:
                self.amt-=new
                print("Remaining balance is = ",self.amt)
        except InsufficientBalanceError as e:
             print(e)
    



        
bank=Bank(3000)
bank.withdraw(20000)
    
```

```python
# 23. Create a function that checks password length and raises a custom exception if length is
# less than 8.
class PassCheckError(Exception):
    pass

class password:
    def check(self,pas):
        if len(pas)>=8:
            print("Password approved")
        else:
            raise PassCheckError("Password length must be >8")
try:
    pas=password()
    passw=input("Enetr your password")
    pas.check(passw)
except PassCheckError as e:
    print(e)
        
```
