#### Data Type & Invalid Values

```python
# 11.The Age column contains numbers as well as values like "unknown" and
# "twenty five". Clean this column and convert it into a proper numeric column.
df["age"]=pd.to_numeric(df["age"],errors="coerce")
age_med=df["age"].median()
df["age"]=df["age"].fillna(age_med)
df["age"]
```

```python
# 12.The Annual Income column contains values such as 650000 INR, ₹720000,
# 1.2L, and N/A. Clean the column and convert the valid values into numeric
# format.
df["annual_income"]=pd.to_numeric(df["annual_income"],errors="coerce")

anu_med=df["annual_income"].median()
df["annual_income"]=df["annual_income"].fillna(anu_med)
df["annual_income"]
```

```python
# 13.Clean the Credit Score column and identify values that cannot be converted
# into numbers.
df["credit_score"]=pd.to_numeric(df["credit_score"],errors="coerce")

cred_med=df["credit_score"].median()
df["credit_score"]=df["credit_score"].fillna(cred_med)
df["credit_score"]
```

```python
# 14.The Loan Amount column contains currency symbols, text, and numeric
# values. Remove the unwanted characters and convert it into numeric format.

df["loan_amount_(inr)"]=pd.to_numeric(df["loan_amount_(inr)"],errors="coerce")
df["loan_amount_(inr)"].replace({"INR":"","₹":"","$":""},inplace=True)
```

```python
df["loan_amount_(inr)"]
```

```python
# 15.The Monthly Visits column contains values such as "five", "N/A", and "-".
# Convert valid values into numeric values and handle invalid values.
import numpy as np
df["monthly_visits"].replace({"N/A":np.nan,"five":5,"-":""})
df["monthly_visits"]=pd.to_numeric(df["monthly_visits"],errors="coerce")
```

```python
df["monthly_visits"]
```

```python
# 16.Clean the Debt to Income column where some values are written like 35%
# and 0.4x. Convert the valid values into a consistent numeric format.
df["debt_to_income"]=pd.to_numeric(df["debt_to_income"],errors="coerce")
df["debt_to_income"]
```

```python
# 17.After cleaning the numerical columns, check which columns are still stored as
# object data type.
df.select_dtypes(include=["int","float"])
df.info()
```

```python
# Categorical Cleaning
```

```python
# 18.Find all unique values in the Employment-Type column.
df["employment_type"].unique()
```

```python
# 19.Standardize the Employment-Type column so that values such as salaried,
# SALARIED, and Salaried become one consistent value.
df["employment_type"].replace({"Self employed":"Self-Employed","self-employed":"Self-Employed","salaried":"Salaried","SALARIED":"Salaried"},inplace=True)
df["employment_type"].str.strip()
```

```python
# 20.Clean the Education Level column and make values such as graduate,
# GRADUATE, and Graduate consistent.
df["education_level"].unique()
df["education_level"].replace({"GRADUATE":"Graduate","graduate":"Graduate","Postgraduate":"Post Graduate","postgraduate":"Post Graduate"},inplace=True)
df["education_level"].str.strip()
```

```python
# 21.Clean the City column. Identify incorrect or inconsistent values such as
# hyderbad, Hyd, Bangalore, and Bengaluru.

df["city"].replace({"MUMBAI":"Mumbai",
                   "mumbai ":"Mumbai",
                   "hyderbad":"Hyderabad",
                   "Hyd":"Hyderabad",
                   "Bangalore":"Bengaluru",
                    " Bengaluru ":"Bengaluru"
                   },inplace=True)
df["city"].str.strip().replace(" ","")
df["city"]
```

```python
df["city"].unique()
```

```python
# 22.Find all unique values in the Customer Segment column and identify invalid
# categories.
df["customer_segment"].unique()
df["customer_segment"].replace({"mass":"Mass"," PREMIUM":"Premium","premium ":"Premium",
                               },inplace=True)
df["customer_segment"]
```

```python
df["customer_segment"].unique()
```

```python
# 23.Clean the Loan Type column so that values such as personal, PERSONAL,
# and personal loan are standardized.
df["loan_type"].unique()
df["loan_type"].replace({"Home Loan":"Home ",
                        "auto ":"Auto",
                        "personal loan":"Personal ",
                         "PERSONAL":"Personal",
                         "Personal ":"Personal",
                         "Home ":"Home"
                        },inplace=True)
df["loan_type"].str.strip()
```

```python
df["loan_type"].unique()
```

```python
# 24.Clean the Loan Approved column where values include Yes, yes, Y, 1, No,
# NO, N, and 0. Convert them into only Yes/No.
df["loan_approved?"].unique()
df["loan_approved?"].replace({"0":"No",
                             "N":"No",
                              "NO":"No",
                              "yes":"Yes",
                              "1":"Yes",
                              "Y":"Yes"
                             },inplace=True)
```

### Outliers & Data Validation

```python
# 25.Check the Age column and identify unrealistic ages such as negative ages or
# ages above 100.
df["age"]=["Unrealistic Age" if i<0 or i>100 else i for i in df["age"]]
df["age"]
```

```python
# 26.Check the Credit Score column and identify values outside the valid credit-
# score range.
df["credit_score"]=["Invalid Range" if i<300 or i>900 else i for i in df["credit_score"]]
df["credit_score"]
```

```python
# 27.Detect outliers in Annual Income using the IQR method.
df["annual_income"]=pd.to_numeric(df["annual_income"],errors="coerce")
q1=df["annual_income"].quantile(0.25)
q3=df["annual_income"].quantile(0.75)
IQR=q3-q1
df[(df["annual_income"]<(q1-1.5*IQR)) | (df["annual_income"] > (q3+1.5*IQR))]
```

```python
# 28.Detect outliers in Loan Amount using the IQR method.
df["loan_amount_(inr)"]=pd.to_numeric(df["loan_amount_(inr)"],errors="coerce")
q1=df["loan_amount_(inr)"].quantile(0.25)
q3=df["loan_amount_(inr)"].quantile(0.75)
IQR=q3-q1
df[(df["loan_amount_(inr)"]<(q1-1.5*IQR)) | (df["loan_amount_(inr)"]>(q3+1.5*IQR))]
```

```python
# 29.Check the Debt to Income column and identify impossible values such as
# negative values or values greater than 1.
df["debt_to_income"]=pd.to_numeric(df["debt_to_income"],errors="coerce")
df["debt_to_income"]=["Impossible value" if i<0 or i>1 else i for i in df["debt_to_income"]]
df["debt_to_income"]
```
