#### Hybrid Inheritance Questions (1-10)

```python
# 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
# from both B and C).
# Concept
# Hybrid Inheritance (Combination of Single + Multiple Inheritance)
# What to do
# • Create parent class A.
# • Create class B inheriting from A.
# • Create another class C.
# • Create class D inheriting from both B and C.
# How to solve
# 1. Create class A.
# 2. Add one method.
# 3. Create class B(A).
# 4. Create class C.
# 5. Create class D(B, C).
# 6. Create an object of D.
# 7. Call methods from all parent classes.

class A:
    def a_info(self):
        print("This is A")
class B(A):
    def b_display(self):
        print("This is class B")
class C:
    def c_info(self):
        print("This is C class")
class D(B,C):
    def display(self):
        print("Class D")

d=D()
d.display()
d.c_info()
d.b_display()
d.a_info()
```

```python
# Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
# Concept
# Real-world Hybrid Inheritance
# What to do
# • Person is the base class.
# • Employee inherits Person.
# • Student inherits Person.
# • TeachingAssistant inherits Employee and Student.
# How to solve
# Hierarchy
# Person
# / \
# Employee Student
# \ /
# TeachingAssistant
# Steps
# 1. Store name in Person.
# 2. Employee stores salary.
# 3. Student stores marks.
# 4. TeachingAssistant stores department.
# 5. Display all information.

class Person:
    def __init__(self, name, **kwargs):
        super().__init__(**kwargs)
        self.name = name


class Employee(Person):
    def __init__(self, salary, **kwargs):
        super().__init__(**kwargs)
        self.salary = salary


class Student(Person):
    def __init__(self, marks, **kwargs):
        super().__init__(**kwargs)
        self.marks = marks


class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, dept):
        super().__init__(name=name, salary=salary, marks=marks)
        self.dept = dept

    def display(self):
        print("Name       =", self.name)
        print("Salary     =", self.salary)
        print("Marks      =", self.marks)
        print("Department =", self.dept)


tech = TeachingAssistant("Vaishnavi", 230000, 67, "Maths")
tech.display()
    
```

```python
# 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
# Concept
# Method Resolution Order (MRO)
# What to do
# Create
# A
# |
# B

# C

# D(B,C)
# Print
# D.mro()
# How to solve
# 1. Create four classes.
# 2. Give same method name in multiple classes.
# 3. Print ClassName.mro().
# 4. Observe Python's searching order.
class A:
    def ainfo(self):
        print("Class A")
class B(A):
    def binfo(self):
        print("Class B")
class C:
    def cinfo(self):
        print("Class C")
class D(B,C):
    def dinfo(self):
        print("Class D")
d=D()
d.ainfo()
d.binfo()
d.cinfo()
d.dinfo()
print(D.__mro__)
```

```python
# 5. Create class A, B, C, D and solve the Diamond Problem using super().
# Concept

# Sumit Sir Notes
# Diamond Problem
# Hierarchy
# A
# / \
# B C
# \ /
# D
# What to do
# Use super() everywhere.
# How to solve
# 1. Create constructor in A.
# 2. Call super() in B.
# 3. Call super() in C.
# 4. Call super() in D.
# 5. Create object of D.
# 6. Observe A constructor runs only once.
class A:
    def __init__(self,a,**kwargs):
        super().__init__(**kwargs)
        self.a=a
        
class B(A):
    def __init__(self,b,**kwargs):
        super().__init__(**kwargs)
        self.b=b
            
class C(A):
    def __init__(self,c,**kwargs):
        super().__init__(**kwargs)
        self.c=c
       
class D(B,C):
    def __init__(self,a,b,c,d):
        super().__init__(a=a, b=b, c=c)
        self.d=d        
    
    def display(self):
        print("A= ",self.a)
        print("B= ",self.b)
        print("C= ",self.c)
        print("D= ",self.d)
        
    
d=D(2,3,4,5)
d.display()

print(D.__mro__)
```

```python
# 11. What is Duck Typing? Explain with a simple example.
# Concept
# Duck Typing
# "If an object behaves like a duck, Python treats it like a duck."
# What to do
# Create two unrelated classes having the same method.
# How to solve
# 1. Create Dog.
# 2. Create Robot.
# 3. Both have speak().
# 4. Write one function calling obj.speak().
# 5. Pass both objects.
class Dog:
    def speak(self):
        print("Dog barks")
class Robot:
    def speak(self):
        print("Robot may speak human language")
def sound(animal):
    animal.speak()
d=Dog()
r=Robot()
sound(d)
sound(r)
```

```python
# 19. Create a hybrid inheritance structure and apply Duck Typing.
# Concept
# Hybrid + Duck Typing

# Sumit Sir Notes
# What to do
# Create hybrid inheritance.
# Create another unrelated class with same method.
# How to solve
# 1. Create hybrid hierarchy.
# 2. Final child has method show().
# 3. Another unrelated class also has show().
# 4. One function calls show() for both.
class A:
    def info(self):
        print("class A")
class B(A):
    def info(self):
        print("class B")
class C:
    def info(self):
        print("class C")
class D(B,C):
    def info(self):
        print("class D")

def display(obj):
    obj.info()
a=A()
b=B()
c=C()
d=D()
display(a)
display(b)
display(c)
display(d)
```

```python
# 20. Create Employee hierarchy using Hybrid Inheritance and use Duck Typing.
# Concept
# Hybrid Inheritance + Duck Typing
# What to do
# Hierarchy
# Person
# |
# Employee

# Manager

# HR

# TeamLead
# Another class
# Freelancer
# Both TeamLead and Freelancer have
# work()
# How to solve

# Sumit Sir Notes
# 1. Build hybrid inheritance.
# 2. Final child defines work().
# 3. Freelancer also defines work().
# 4. Write a function:
# def do_work(obj):
# obj.work()
# 5. Pass both objects to the function.
# 6. Observe that the same function works for unrelated classes because of Duck Typing.

class Person:
    def __init__(self,name,**kwargs):
        super().__init__(**kwargs)
        self.name=name
class Employee(Person):
    def __init__(self,age,**kwargs):
        super().__init__(**kwargs)
        self.age=age
class Manager(Employee):
    def __init__(self,salary,**kwargs):
        super().__init__(**kwargs)
        self.salary=salary
class HR(Employee):
    def __init__(self,dept,**kwargs):
        super().__init__(**kwargs)
        self.dept=dept
class TeamLead(Manager,HR):
    def __init__(self,name,age,salary,dept):
        super().__init__(name=name,age=age,salary=salary,dept=dept)
      
        
    def work(self):
        print("Team lead works in office")
    def display(self):
        print("Name= ",self.name)
        print("Salary= ",self.salary)
        print("Dept= ",self.dept)
class Freelancer:
    def work(self):
        print("Freelancer works anywhere")

def job(obj):
    obj.work()

team=TeamLead("vaishnavi",25,50000000,"AI ML")
free=Freelancer()
    
team.display()
print()
job(team)
job(free)
```
