```python
#Write a lambda function to return the square of a number.
square=lambda x: x*x
square(5)
```

```python
#Write a lambda function to return the cube of a number.
cube=lambda x :x*x*x
cube(4)
```

```python
#Write a lambda function that returns "Even" if a number is even; otherwise "Odd".
even=lambda x :"Even" if x%2==0 else "odd"
even(34)
```

```python
#9.Sort a list of numbers in ascending order using lambda.

lis=[34,56,2,4,8,0,10]
asc=sorted(lis,key=lambda x:x)
asc
```

```python
#Sort a list of numbers in descending order using lambda.
lis=[34,56,2,4,8,0,10]
des=sorted(lis,key=lambda x:x,reverse=True)
des
```

```python
#Use reduce() with lambda to find the sum of all numbers.
from functools import reduce
lis=[34,56,2,4,8,0,10]
summ=reduce(lambda x,y : x+y,lis)
print(summ)
```

```python
#Use reduce() with lambda to find the product of all numbers.
from functools import reduce
lis=[34,56,2,4,8,0,10]
product=reduce(lambda x,y:x*y,lis)
print(product)
```

```python
#Given:
students = [
{"name":"Ram","marks":[70,80,90]},
{"name":"Amit","marks":[90,92,94]},
{"name":"Neha","marks":[60,65,70]},
{"name":"Raj","marks":[88,84,90]}
]
#Using only lambda with built-in functions (map, filter, sorted, max, min, reduce where appropriate):
#1. Calculate each student's average marks.
#2. Sort students by average marks (highest first).
#3. Find the topper.
#4. Filter students whose average is greater than 80.
#5. Print only the names of the filtered studen

avg_marks=list(map(lambda x:{ "name": x["name"],
                            "Average": sum(x["marks"]) /len (x["marks"]) },students))
avg_marks
```

```python
sort_marks = sorted(avg_marks,key =lambda x: x["Average"],reverse = True)
sort_marks
```

```python
topper= max(sort_marks,key =lambda x: x["Average"])
topper
```

```python
std_filter=list(filter(lambda x: x["Average"]>=80, avg_marks))
print(std_filter)
```

```python
des_sort=sorted(avg_mark,key=lambda s: s["average"] ,reverse=True )
des_sort
```

```python
topper=max(avg_mark,key=lambda s:s["average"])
topper
```

```python
low=min(avg_mark,key=lambda s:s["average"])
low
```

```python
condition=list(filter(lambda s: s["average"]>90 ,avg_marks))
condition
```
