```python
# 1. Create two parent classes Father and Mother. Create a child class Child that inherits from
# both.
class Father:
    def __init__(self,money,**kwargs):
        super().__init__(**kwargs)
        self.money=money
class Mother:
    def __init__(self,beauty,**kwargs):
        super().__init__(**kwargs)
        self.beauty=beauty
class Child(Father,Mother):
    def __init__(self,money,beauty,smart):
        super().__init__(money=money,beauty=beauty)
        self.smart=smart
    def display(self):
        print("The father has = ",self.money)
        print("The mother has = ",self.beauty)
        print("The child has = ",self.smart)

c=Child("Money","Beauty","Smartness")
c.display()
```

##### Multiple Inheritance with Abstract Methods (19-25)

```python
# 19. Create two abstract classes Shape and Color. Create a concrete class that inherits from
# both and implements abstract methods.
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def draw(self):
        pass
class Color(ABC):
    @abstractmethod
    def paint(self):
        pass
class concrete(Shape,Color):
    def draw(self):
        print("Draw you shape")
    def paint(self):
        print("color it")

con=concrete()
con.draw()
con.paint()
    
```

```python
# 20. Define abstract class Payment and abstract class Notification. Create class
# OnlinePayment that inherits from both.
class Payment(ABC):
    @abstractmethod
    def amount(self):
        pass
class Notification(ABC):
    @abstractmethod
    def app(self):
        pass
class OnlinePayment(Payment,Notification):
    def __init__(self,Amount,App):
        self.Amount=Amount
        self.App=App
    def amount(self):
        print("Enter the amount= ",self.Amount)
    def app(self):
        print("Enter the app=  ",self.App)
online=OnlinePayment(2000,"Gpay")
online.amount()
online.app()
```

```python
# 27. (Problem) Create abstract class Device with abstract method power_on() and class
# Network with method connect(). Create class SmartPhone that inherits from both and
# implements the abstract method.
from abc import ABC, abstractmethod
class Device(ABC):
    @abstractmethod
    def power_on(self):
        pass

class Network(ABC):
    
    def connect(self):
        print("The power is connected")
class SmartPhone(Device,Network):
    def power_on(self):
        print("The device is starting")
    def show(self):
        self.power_on()
        self.connect()
        

smart=SmartPhone()
smart.show()
```

```python
# 28. (Problem) Design a system using multiple inheritance:
# • Class Writer
# • Class Editor Create class Author that inherits from both. Add static method to count
# total authors.
class Writer:
    def write(self):
        print("Writing the book")
    
class Editor:
    def edit(self):
        print("Editing the book")
class Author(Writer,Editor):
    total_count=0
    
    def __init__(self,name):
        self.name=name
        Author.total_count+=1
    @staticmethod
    def count():
        print("The total authors = ",Author.total_count)

A1=Author("vaishnavi")
A2=Author("Mangesh")
A1.count()
```
