## Seaborn catplot() — 30 Questions

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# 1. Survival by Gender
# Use catplot() to show the number of passengers who survived and did not survive for each
# gender.
# • X-axis → sex
# • Hue → survived
# • Chart → count

titanic=sns.load_dataset("titanic")
titanic
sns.catplot(data=titanic,
           x="sex",
           hue="survived",
           kind="count",
           palette="deep")
plt.title("Survival by Gender")
plt.show()
```

```python
# 2. Survival by Passenger Class
# Create a categorical plot showing the number of passengers in each class.
# Use:
# • x="class"
# • kind="count"
sns.catplot(
    data=titanic,
    x="class",
    kind="count",
    palette="pastel"
)
plt.title("Survival by Passenger Class")
plt.show()
```

```python
# 3. Survival Count by Class
# Show the number of survivors and non-survivors for each passenger class.
# Use:
# x="class"
# hue="survived"
# kind="count"
sns.catplot(
    data=titanic,
    x="class",
    hue="survived",
    kind="count",
    palette="pastel"
)

plt.title("Survival Count by Class")
plt.show()
```

```python
# 4. Survival Percentage by Gender
# Create a categorical visualization comparing survival across sex.
# Use catplot() and choose an appropriate categorical plot.
# Then identify which gender has a higher survival rate.
sns.catplot(
    data=titanic,
    x="survived",
    hue="sex",
    palette="deep",
    kind="boxen"
)
plt.title("Survival Percentage by Gender")
plt.show()
```

```python
# 5. Age Distribution by Gender
# Use catplot() to compare passenger age across different genders.
# Try an appropriate kind for showing the distribution

sns.catplot(
    data=titanic,
    x="sex",
    y="age",
    hue="sex",
    palette="pastel",
    kind="box"
)
plt.title("Age Distribution by Gender")
```

```python
# 6. Age Distribution by Passenger Class
# Create a categorical plot showing age distribution for:
# • First class
# • Second class
# • Third class

sns.catplot(
    data=titanic,
    x="class",
    y="age",
    hue="sex",
    palette="pastel",
    kind="box"
)
plt.title("Age Distribution by Passenger Class")
plt.show()
```

```python
# 7. Fare by Passenger Class
# Use catplot() to compare fare across the three passenger classes.
# Add:
# • Appropriate color
# • Title
# • X/Y labels

sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    palette=["pink",'purple'],
    kind="box"
)
plt.title("Fare by Passenger Class")
plt.show()
```

```python
# 8. Fare Distribution by Gender
# Create a categorical visualization of fare for male and female passengers.
# Identify whether the fare distributions appear different.
sns.catplot(
    data=titanic,
    x="sex",
    y="fare",
    hue="sex",
    palette=["red","purple"],
    kind="boxen"
)
plt.title("Fare Distribution by Gender")
plt.show()
```

```python
# 12. Passenger Class vs Fare with Gender
# Create a categorical visualization comparing fare across passenger classes.
# Use:
# x="class"
# y="fare"
# hue="sex"
# Choose an appropriate kind.

sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    kind="bar",
    palette=["pink","purple"]
)
```

```python
# 13. Survival vs Age Group
# Create an age_group column:
# • Child → Age < 18
# • Adult → 18–59
# • Senior → 60+
# Use catplot() to compare survival across these groups.
titanic["age_group"]=titanic["age"].apply(lambda x: "Child" if x<18 else "Adult" if x<60 else "Senior")

sns.catplot(
    data=titanic,
    x="age_group",
    y="survived",
    kind="bar",
    palette='deep'
    
)
plt.title("Survival vs Age Group")
plt.show()
```

```python
# 14. Class, Gender and Survival
# Create a categorical visualization where:

# Sumit Sir Notes
# • X → class
# • Y → survival-related measure
# • Hue → sex
# • Columns → survived
# Use col= to create separate panels.
sns.catplot(
    data=titanic,
    x="class",
    y="survived",
    hue="sex",
    col="survived",
    palette=["hotpink",'purple'],
    kind="bar"
)

plt.title("Class, Gender and Survival")
plt.show()
```

```python
# 15. Titanic Multi-Panel Analysis
# Create a catplot() with:
# • X → class
# • Y → fare
# • Hue → sex
# • Columns → survived
# Use an appropriate categorical kind.
# Add:
# • Figure size
# • Title
# • Color palette
# • Appropriate labels
# Explain what you observe.

sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="bar",
    palette=["red","purple"],
    
)
plt.title("Titanic Multi-Panel Analysis")
plt.show()
```

```python
# 30. Final Seaborn catplot() Challenge
# Using the Tips dataset, create a professional categorical visualization with:
# x="day"
# y="total_bill"
# hue="sex"
# col="time"
# Choose the most appropriate kind.
# Customize:
# • palette
# • height
# • aspect
# • col_wrap if appropriate
# • Title
# • Axis labels
# • Legend
# Then answer:
# 1. Which day has the highest typical bill?
# 2. Is there a difference between male and female customers?
# 3. Does the pattern change between lunch and dinner?
# 4. Which combination of day and time appears to have the highest bills?

sns.catplot(
    data=tips,
    x="day",
y="total_bill",
hue="sex",
col="time",
    palette="deep",
    height=8,
    aspect=0.8,
    kind="strip",
    linewidth=0.9
    
)
plt.title("day vs bill")
plt.xlabel("Day")
plt.ylabel("Bills")
plt.legend(["bill"])
```
