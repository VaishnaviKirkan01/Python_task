```python
# 11. Create class BankAccount with private __balance. Add deposit and withdraw
# methods. Raise custom exception if balance becomes negative.
class NegativeValueError(Exception):
    pass
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance
    def get_marks(self):
        return f"The Balance are = {self.__balance}"
    def set_marks(self,new):
        if new>=0:
            self.__balance+=new
            print("Updated balance = ",self.__balance)
        else:
            raise NegativeValueError("Entered negative value")
    def deposit(self,amt):
        if amt>=0:
            self.__balance+=amt
            print("The amount after deposited is= ",self.__balance)
        else:
            raise NegativeValueError("Entered negative value")
    def withdraw(self,with_amt):
        if with_amt>=0:
            self.__balance-=with_amt
            print("The amount after withdrawn is= ",self.__balance)
        else:
            raise NegativeValueError("Entered negative value")

try:
    b1=BankAccount(50000)
    print("Current Balance =",b1.get_marks())
    b1.deposit(200)
    b1.withdraw(3000)
    b1.withdraw(4000)
except NegativeValueError as e:
    print(e)
```

#### Inheritance + Abstraction + super() (15-19)

```python
# 15. Create single level inheritance: Person → Student. Use super() in constructor.
class Person:
    def __init__(self,name):
        self.name=name
class Student(Person):
    def __init__(self,name,marks):
        super().__init__(name)
        self.marks=marks
    def display(self):
        print("Name= ",self.name)
        print("Marks= ",self.marks)
std=Student("Vaishnavi","99")
std.display()
```

```python
# 16. Create multi-level inheritance: Person → Employee → Manager. Override a
# method at each level.
class Person:
    def __init__(self,name,**kwargs):
        super().__init__(**kwargs)
        self.name=name
class Employee(Person):
    def __init__(self,name,em_id,**kwargs):

        super().__init__(name,**kwargs)
        self.em_id=em_id 
class Manager(Employee):
    def __init__(self,name,em_id,salary):
        super().__init__(name,em_id)
        self.salary=salary
    def display(self):
        print("Name= ",self.name)
        print("Em_id= ",self.em_id)
        print("Salary= ",self.salary)
man=Manager("Vaishnavi","Emp123",20000000)
man.display()
```

```python
# 17. Create hierarchical inheritance: Parent Animal → Dog and Cat. Override sound()
# method.
class Animal:
    def sound(self):
        print("Animal make sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat Meows")
d=Dog()
c=Cat()
d.sound()
c.sound()
```

```python
# 18. Create abstract class Shape with abstract method area(). Implement it in Circle
# and Square using super().
from abc import ABC,abstractmethod
class Shape(ABC):
    def __init__(self,value):
        self.value=value
        
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,r):
        super().__init__(r)
    def area(self):
        area=3.14*self.value**2
        print("Area= ",area)
class Square(Shape):
    def __init__(self,s):
        super().__init__(s)
    def area(self):
        area=self.value**2
        print('Area= ',area)

c=Circle(34)
s=Square(6)
c.area()
s.area()
```

```python
# 19. Create multiple inheritance with two classes Father and Mother. Create child
# class and resolve method conflict.
class Father:
    def show(self):
        print("This is Father's method.")
class Mother:
    def show(self):
        print("This is Mother's method.")
class Child(Father, Mother):
    def show(self):
        print("This is Child's method.")
        Father.show(self)   
        Mother.show(self)   
c = Child()
c.show()
```

```python
# 21. Create class Point and overload + and > operators.
class Point:
    def __init__(self,a):
        self.a=a
    def __add__(self,other):
        return self.a+other.a
    def __gt__(self,other):
        return self.a>other.a
p1=Point(3)
p2=Point(4)
print(p1+p2)
print(p1>p2)
```
