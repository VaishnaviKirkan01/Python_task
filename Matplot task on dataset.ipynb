```python
# 1. Plot monthly Revenue as a line chart. Add title, axis labels, markers, grid, and legend.
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.read_csv("C:\\Users\\vaish\\Downloads\\matplotlib_sales_clean_1000.csv")

x=df["Revenue"]
plt.plot(x,
         linewidth=1,
        marker="*",
         markeredgecolor="pink"
        )
plt.legend(["Revenue"])
plt.title("Monthly Revenue")
plt.grid(True,alpha=0.9)
plt.show()
```

```python
# 13. Create a stacked bar chart showing Revenue by month for each Product_Category.
df["Month"]=pd.to_datetime(df["Order_Date"]).dt.month
x=df.groupby(["Month","Product_Category"])["Revenue"].sum().unstack()
x.plot(kind="bar",stacked=True)
plt.show()
```

```python
# 14. Create a grouped bar chart comparing Revenue across cities for different customer
# segments.
x=df.groupby(["City","Customer_Segment"])["Revenue"].sum().unstack()
x.plot(kind="bar",stacked=True)
plt.show()
```

```python
# Plot monthly Revenue and Cost together and shade the area between the two lines
# to represent Profit.
df["month"]=pd.to_datetime(df["Order_Date"]).dt.month
x=df.groupby("Month")["Revenue"].sum()
y=df.groupby("Month")["Cost"].sum()
plt.plot(x)
plt.plot(y)
plt.fill_between(x.index,x,y)
```

```python
# 16. Find the month with the highest Revenue and annotate that point on the line chart.
df["Month"]=pd.to_datetime(df["Order_Date"]).dt.month
x=df.groupby("Month")["Revenue"].sum()
plt.plot(x)
m=x.idxmax()
r=x.max()

plt.annotate("Highest",xy=(m,r),xytext=(m,r))
```

```python
# 17. Find the month with the lowest Profit and annotate it on the visualization.

df["Month"]=pd.to_datetime(df["Order_Date"]).dt.month
x=df.groupby("Month")["Revenue"].sum()
plt.plot(x)
m=x.idxmin()
r=x.min()
plt.annotate("Lowest",xy=(m,r),xytext=(m,r))
```

```python
# 18. Create a line chart of monthly Revenue and calculate a 3-month moving average.
# Plot both lines togeth
x=df.groupby("Month")["Revenue"].sum()
plt.plot(x)
x1=x.rolling(3).mean()
plt.plot(x1)
```
