#### Method Overloading Questions (1-7)

```python
# 1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method
# name.
# Concept
# Method Overloading (Python uses default arguments or *args to achieve overloading.)
# What to do
# • Create a class Calculator.
# • Create one method add().
# • The method should work for either 2 numbers or 3 numbers.
# How to solve
# 1. Create class Calculator.
# 2. Define method add().
# 3. Use default arguments or *args.
# 4. Check how many numbers are provided.
# 5. Return the sum.

class Calculator:
    def add(self,a,b,c=0,d=0):
        result= a+b+c+d
        print(result)
cal=Calculator()
cal.add(4,5)
cal.add(4,5,6)
```

```python
# 2. Create an Area class that calculates the area of a square and rectangle using the same
# method.
# Concept
# Method Overloading using default arguments.
# What to do
# • One parameter → Square.
# • Two parameters → Rectangle.
# How to solve
# 1. Create class Area.
# 2. Define method calculate().
# 3. If only one value is provided, calculate the square's area.
# 4. If two values are provided, calculate the rectangle's area.
# 5. Return the result.
class Area:
    def calculate(self,length,width=0):
        if width==0:
            square=length*length
            return f"Square area= {square}"
        else:
            rectangle=length*width
            return f"Rectangle area= {rectangle}"
area=Area()
area.calculate(5)
area.calculate(8,5)     
    
```

```python
# 4. Create a MathOperations class that multiplies any number of values.
# Concept
# Method Overloading using *args
# What to do
# Multiply:
# • 2 numbers
# • 3 numbers
# • 4 numbers
# • More numbers
# How to solve
# 1. Create class MathOperations.
# 2. Define method multiply(*args).
# 3. Initialize the answer as 1.
# 4. Loop through all numbers.
# 5. Multiply each number.
# 6. Return the final product.

class MathOperations:
    def multiply(self,*args):
        result=1
        for i in args:
            result=result*i
            
        return result
            
math=MathOperations()
math.multiply(3,3)
math.multiply(3,3,6)
```

```python
# 5. Create a Display class that prints different messages based on the argument type.
# Concept
# Method Overloading + Type Checking
# What to do
# Accept:
# • Integer
# • String
# • List
# How to solve
# 1. Create class Display.
# 2. Create method show().
# 3. Use isinstance().
# 4. Print different messages depending on the object's type.
class Display:
    def show(self,x):
        if isinstance(x,int):
            print("x=integer")
        elif isinstance(x,str):
            print("x=String")
        elif isinstance(x,list):
            print("x=List")
dis=Display()
dis.show(3)
```

```python
# 6. Create a Volume class that calculates the volume of a cube and cuboid.
# Concept
# Method Overloading
# What to do
# • One parameter → Cube
# • Three parameters → Cuboid
# How to solve
# 1. Create class Volume.
# 2. Create method calculate().
# 3. If one value is given, calculate the cube's volume.
# 4. Otherwise, calculate the cuboid's volume.
class Volume:
    def calculate(self,length,width=0,height=0):
        if width!=0:
            if height!=0:
                cubiod=length*width*height
                return f"CUBIOD AREA = {cubiod}"
                
        else:
            cube=length**3
            return f"CUBE AREA = {cube}"
vol=Volume()
vol.calculate(3,4,5)
```

```python
# 7. Create a PrintData class that prints one or many values.
# Concept
# Method Overloading using *args
# What to do
# Print:
# • One value
# • Multiple values
# How to solve
# 1. Create class PrintData.
# 2. Create method print_data(*args).
# 3. Loop through the arguments.
# 4. Print each value.
class PrintData:
    def print_data(self,*args):
        res=1
        for res in args:
            print(res)
pri=PrintData()
pri.print_data(23,45)
```

#### Method Overriding Questions (8-14)

```python
# 8. Create an Animal class with sound(). Override it in Dog and Cat.
# Concept
# Method Overriding
# What to do
# • Animal → Generic sound
# • Dog → Bark
# • Cat → Meow
# How to solve
# 1. Create parent class Animal.
# 2. Define method sound().
# 3. Inherit Dog and Cat.
# 4. Override sound().
# 5. Create objects and call the method.
class Animal:
    def sound(self):
        print("Animal make sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks= bow bow")
class Cat(Animal):
    def sound(self):
        print("Cat meows= meow meow")
dog=Dog()
cat=Cat()
dog.sound()
cat.sound()
```

```python
# 9. Create Vehicle, Car, and Bike classes.
# Concept
# Method Overriding
# What to do
# Override the start() method.
# How to solve
# 1. Create Vehicle.
# 2. Add method start().
# 3. Create Car.
# 4. Create Bike.
# 5. Override start() in both child classes.
class Vehicle:
    def start(self):
        print("The vehicle has started")
class Car(Vehicle):
    def start(self):
        print("Car has started")
class Bike(Vehicle):
    def start(self):
        print("Bike has started")
car=Car()
bike=Bike()
car.start()
bike.start()
```

```python
# 11. Create Shape, Circle, and Rectangle classes.
# Concept
# Method Overriding
# What to do
# Override the area() method.
# How to solve
# 1. Create Shape.
# 2. Add area().
# 3. Inherit Circle.
# 4. Inherit Rectangle.
# 5. Override area() in each child class.
class Shape:
    def area(self):
        print("The shapes has area")
class Circle(Shape):
    def area(self,r):
        print("The area of circle is = ",3.14*r)
class Rectangle(Shape):
    def area(self,l,w):
        print("The area of rectangle is = ",2*(l+w))

cir=Circle()
rect=Rectangle()
cir.area(4)
rect.area(4,5)
```

```python
# 12. Create BankAccount, SavingsAccount, and CurrentAccount.
# Concept
# Method Overriding
# What to do
# Override the interest() method.
# How to solve
# 1. Create BankAccount.
# 2. Define interest().
# 3. Override it in both child classes.
# 4. Print different interest rates.

class BankAccount:
    def __init__(self,amount):
        self.amount=amount
class Current(BankAccount):
    def __init__(self,amount):
        BankAccount.__init__(self,amount)   
    def display(self):
        self.intrest=self.amount*0.02
        print("Current account intrest= ",self.intrest)
class Savings(BankAccount):
    def __init__(self,amount):
        BankAccount.__init__(self,amount)   
    def display(self):
        self.intrest=self.amount*0.05
        print("Savings account intrest= ",self.intrest)    
curr=Current(400)
curr.display()
save=Savings(400)
save.display()
```

```python
# 13. Create Parent and Child classes using super().
# Concept
# Method Overriding + super()
# What to do
# Override display().
# How to solve
# 1. Create Parent.
# 2. Add display().
# 3. Create Child.
# 4. Override display().
# 5. Call super().display().
# 6. Print additional information.
class Parent:
    def __init__(self,name,age,**kwargs):
        super().__init__(**kwargs)
        self.name=name
        self.age=age
        
    def display(self):
        print("Name=  ",self.name)
        print("Age=   ",self.Age)
class Child(Parent):
    def __init__(self,name,age,contact,address):
        super().__init__(name=name,age=age)
        self.contact=contact
        self.address=address
    def display(self):
        print("Contact = ",self.contact)
        print("Address = ",self.address)
child=Child("pari",4,34567,"hyd")
child.display()
```

#### Operator Overloading Questions (15-20)

```python
# 15. Create a Point class and overload the + operator.
# Concept
# Operator Overloading (__add__)
# What to do
# Add two points together.
# How to solve
# 1. Create class Point.
# 2. Store x and y coordinates.
# 3. Overload __add__().
# 4. Return a new point with added coordinates.
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return Point(self.x+other.x, self.y+other.y)
    def display(self):
        print(f"{self.x} ,{self.y}")
p1=Point(3,4)
p2=Point(6,7)
p3= p1+p2
p3.display()
```

```python
# 16. Create a Book class and overload the + operator.
# Concept
# Operator Overloading
# What to do
# Add the number of pages of two books.
# How to solve
# 1. Create class Book.
# 2. Store pages.
# 3. Overload __add__().
# 4. Return the total pages.
class Book:
    def __init__(self,pages):
        self.pages=pages
    def __add__(self,other):
        total=self.pages+other.pages
        return total
        
b1=Book(3)
b2=Book(6)
print(b1+b2)
```

```python
# 17. Create a Student class and overload the > operator.
# Concept
# Comparison Operator Overloading (__gt__)
# What to do
# Compare marks of two students.
# How to solve
# 1. Create class Student.
# 2. Store marks.
# 3. Overload __gt__().
# 4. Compare the marks.
# 5. Return True or False.

class Student:
    def __init__(self,marks):
        self.marks=marks
    def __gt__(self,other):
        result=self.marks>other.marks
        return result
s1=Student(57)
s2=Student(99)
print(s1>s2)
    
```

```python
# 18. Create a Rectangle class and overload the == operator.
# Concept
# Equality Operator Overloading (__eq__)
# What to do
# Compare rectangles using their area.
# How to solve
# 1. Create class Rectangle.
# 2. Store length and width.
# 3. Calculate area.
# 4. Overload __eq__().
# 5. Compare the areas.
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def __eq__(self,other):
        result=self.length*self.width ==other.length*other.width
        return result
rect1=Rectangle(34,45)
rect2=Rectangle(34,45)
print(rect1==rect2)
```

```python
# 19. Create a Money class and overload the + and - operators.
# Concept
# Multiple Operator Overloading
# What to do
# Perform addition and subtraction of money amounts.
# How to solve
# 1. Create class Money.
# 2. Store the amount.
# 3. Overload __add__().
# 4. Overload __sub__().
# 5. Return new Money objects after each operation.

class Money:
    def __init__(self,amount):
        self.amount=amount
    def __add__(self,other):
        result= self.amount+other.amount
        return f"Addition ={result}"
    def __sub__(self,other):
        res=self.amount-other.amount
        return f"Substraction ={res}"
m1=Money(2300)
m2=Money(4600)
print(m1+m2)
print(m1-m2)
```

```python
# 20. Create a Distance class and overload <, >, and ==.
# Concept
# Comparison Operator Overloading
# What to do
# Compare two distance objects.
# How to solve
# 1. Create class Distance.
# 2. Store the distance value.
# 3. Overload:
# o __lt__() for <
# o __gt__() for >
# o __eq__() for ==
# 4. Compare the stored distance values.
# 5. Return True or False.
class Distance:
    def __init__(self,dist):
        self.dist=dist
    def __lt__(self,other):
        result=self.dist<other.dist
        return f"Lesser than {result}"
    def __gt__(self,other):
        result=self.dist>other.dist
        return f"Greater than {result}"
    def __eq__(self,other):
        result=self.dist==other.dist
        return f"Greater than {result}"

d1=Distance(50)
d2=Distance(40)
print(d1<d2)
print(d1>d2)
print(d1==d2)
```
