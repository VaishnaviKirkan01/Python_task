```python
# 1. Class Name: BankAccount
# Problem
# Create a class named BankAccount.
# Explanation
# A bank account stores money. The balance should not be directly accessible from outside
# the class, so create a private attribute named __balance.
# Implement the following methods:
# • deposit(amount)
# o Deposit money into the account.
# o Amount must be greater than 0.
# • withdraw(amount)
# o Withdraw money.
# o Withdrawal should happen only if:
# ▪ amount > 0
# ▪ amount <= available balance

class BankAccount:
    def __init__(self,bal):
        self.__balance=bal
    
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print("money deposited =",amount)
            print("total balance =",self.__balance)
    def withdraw(self,amount):
        if amount>0 and amount<=self.__balance:
            self.__balance-=amount
            print("withdrawn money = ",amount)
            print("remaining amount = ",self.__balance)

bank=BankAccount(2000)
bank.deposit(200)
bank.withdraw(300)
```

```python
# 16. Class Name: Shape
# Problem
# Create an abstract class Shape.
# Explanation
# Declare abstract methods
# • calculate_area()
# • calculate_perimeter()
# Create subclasses like
# • Circle
# • Rectangle
# Implement both methods.

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def calculate_area(self):
        area = 3.14 * self.r * self.r
        print("Area of Circle =", area)

    def calculate_perimeter(self):
        perimeter = 2 * 3.14 * self.r
        print("Perimeter of Circle =", perimeter)


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        area = self.length * self.width
        print("Area of Rectangle =", area)

    def calculate_perimeter(self):
        perimeter = 2 * (self.length + self.width)
        print("Perimeter of Rectangle =", perimeter)


# Driver Code
c = Circle(5)
c.calculate_area()
c.calculate_perimeter()

r = Rectangle(3, 5)
r.calculate_area()
r.calculate_perimeter()


```

```python
# 25. Class Name: Shape3D
# Problem
# Create abstract class Shape3D.
# Explanation
# Abstract methods
# • volume()
# • surface_area()
# Implement in
# • Cube
# • Sphere
# • Cylinder
# Each uses different formulas.
class Shape3D(ABC):
    @abstractmethod
    def volume(self):
        pass
    def surface_area(self):
        pass

class Cube(Shape3D):
    def __init__(self, side):
        self.side = side
    def volume(self):
        vol=self.side**3
        print("cube volume = ",vol)
    def surface_area(self):
        sur=6*self.side**2
        print("surface area cube= ",sur)
        
class Sphere(Shape3D):
    def __init__(self, side):
        self.side = side
    def volume(self):
        vol=4/3*3.14*self.side**2
        print("Sphere volume = ",vol)
    def surface_area(self):
        sur=4*3.14*self.side**2
        print("surface area Sphere= ",sur)
        
class Cylinder(Shape3D):
    def __init__(self, side, height):
        self.side = side
        self.height = height
    def volume(self):
        vol=3.14*self.side*2*self.height
        print("Cylinder volume = ",vol)
    def surface_area(self):
        sur=2*3.14*self.side(self.side+self.height)
        print("surface area Cylinder= ",sur)

cube=Cube(5)
cube.volume()
cube.surface_area()
print()
sphere=Sphere(5)
sphere.volume()
sphere.surface_area()
print()
cylinder=Cylinder('5','7')
cylinder.volume()
cylinder.surface_area()
```

```python
# 30. Class Name: FileHandler
# Problem
# Create abstract class FileHandler.
# Explanation
# Abstract methods
# • read()
# • write(data)
# Implement in
# • TextFileHandler
# • CSVFileHandler
# Each subclass should define how it reads from and writes to its respective file type.
from abc import ABC,abstractmethod
class FileHandler(ABC):
    @abstractmethod
    def read(self):
        pass
    def write(self):
        pass

class TextFileHandler(FileHandler):
    def __init__(self,data):
        self.data=data
    def read(self):
        print("reads textfiles")
    def write(self):
        print(self.data)

class CSVFileHandler(FileHandler):
    def __init__(self,data):
        self.data=data
    def read(self):
        print("reads CSVFile")
    def write(self):
        print(self.data)

text=TextFileHandler("heyyy this is text file")
text.read()
text.write()
print()
csv=CSVFileHandler("heyyy csv file")
csv.read()
csv.write()
```
