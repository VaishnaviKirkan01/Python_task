**26-30: kwargs (Multiple Keyword Arguments)

```python
#26. Write a function student_info(**kwargs) that prints all key-value pairs.
def student_info(**kwargs):
    for key,value in kwargs.items():
        print(key,":",value)

student_info(name="neha",age=45)
    
```

```python
#27. Create a function create_profile(**kwargs) that prints name, age, city, etc.
def create_profile(**kwargs):
    for key,value in kwargs.items():
        print(key,":",value)

create_profile(name="neha",age=34,city="hingoli")
```

```python
#28. Write a function calculate_total(**kwargs) where keys are item names and values are prices.
def calculate_total(**kwargs):
    total=0
    for item,price in kwargs.items():
        print(item,":",price)
        total+=price
    return total

calculate_total(pen=8,paper=6,books=50,pencil=35)
```

```python
#29. Make a function display_info(name, **kwargs) that prints name and other details.
def display_info(name,**kwargs):
    for key,value in kwargs.items():
        print(f"{name} {key} : {value}")

display_info(name="neha",age=23,city="bidar",state="mh")
              
    
```

```python
#30. Write a function save_data(**kwargs) that prints "Saving data:" followed by all key-value pairs.
def save_data(**kwargs):
    for key,value in kwargs.items():
        print(f"Saving data , {key}:{value}")

save_data(name="neha",state="MH",mob_no=234567)
```
