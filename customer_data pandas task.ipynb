```python
# 6. Find the rows where Customer ID is missing.
df["Customer ID"].isna()
```

```python
# 7. Remove leading/trailing whitespace from all column names and convert them
# to lowercase with underscores.
df.columns=df.columns.str.strip().str.lower().str.replace(" ","_")
df.columns
```

```python
# 8. Identify duplicate Customer ID values and count how many duplicates exist.
df["customer_id"].duplicated().sum()
```

```python
# 9. Remove duplicate customer records while keeping the first occurrence.
df["customer_id"].drop_duplicates()
```

```python
# 10.Find all unique values in the customer_segment column and identify
# inconsistent categories such as Consumer, consumer, consumer , etc.
df["customer_segment"].unique()
```

```python
# 11.Clean the customer_segment column by standardizing capitalization and
# whitespace.
df["customer_segment"].str.capitalize().str.strip()
```

```python
# 12.Find all unique values in the gender column and convert them into consistent
# categories such as Male and Female.
df["gender"].unique()
```

```python
df["gender"].replace({"m":"Male","M":"Male","F":"Female","f":"Female"},inplace=True)
```

```python
df["gender"]
```

```python
# 13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and
# No, no, N, 0 become 0.
df["is_active"].replace({"Yes":1,"yes":1,"Y":1,"No":0,"no":0,"N":0},inplace=True)
df["is_active"]
```

```python
# 14.Convert the churn_flag column from categorical values (Yes/No) into a binary
# numerical column (1/0).
df["churn_flag"].replace({"Yes":1,"No":0},inplace=True)
df["churn_flag"]
```

```python
# 15.Identify all values in income_usd that cannot be directly interpreted as
# numbers.
df["income_usd"].astype(str).str.replace(",","").str.replace("$","").str.strip()
```

```python
df["income_usd"]=pd.to_numeric(df["income_usd"],errors="coerce")
df["income_usd"].dtype
```

```python
# 17.Clean the Age column so values such as "35 years" become numeric values.
df["age"]=pd.to_numeric(df["age"],errors="coerce")
```

```python
# 18.Identify unrealistic age values using NumPy/Pandas and replace them with
# NaN.
import numpy as np
df["age"]=[np.nan if i<0 or i>100 else i for i in df["age"]]
df["age"]
```

```python
# 19.Clean the credit_score column by removing text such as "pts" and converting
# the column to numeric.
df["credit_score"]=df["credit_score"].str.replace("pts","")
df["credit_score"]
```

```python
df["credit_score"]=pd.to_numeric(df["credit_score"],errors="coerce")
```

```python
# 20.Detect outliers in income_usd using the IQR method with NumPy/Pandas.
df["income_usd"]=pd.to_numeric(df["income_usd"],errors="coerce")
q1=df["income_usd"].quantile(0.25)
q2=df["income_usd"].quantile(0.75)
IQR=q2-q1
df[(df["income_usd"]<(q1-1.5*IQR)) | (df["income_usd"]>(q2+1.5*IQR))]
```

```python
# 21.Detect outliers in credit_score using the IQR method.
df["credit_score"]=pd.to_numeric(df["credit_score"],errors="coerce")
q1=df["credit_score"].quantile(0.25)
q2=df["credit_score"].quantile(0.75)
IQR=q2-q1
df[(df["credit_score"]<(q1-1.5*IQR)) | (df["credit_score"]>(q2+1.5*IQR))]
```

```python
# 22.Detect suspicious values such as -999 and 99999 in tenure_years and
# satisfaction.
df["tenure_years"]==-999
df["tenure_years"]==99999
```

```python
df["satisfaction"]==-999
df["satisfaction"]==99999
```

```python
# 23.Replace invalid/suspicious numerical values with NaN.
df[["tenure_years",'satisfaction']].replace({-999:np.nan,99999:np.nan})
```

```python
# 24.Fill missing numerical values using the median of each numerical column.
df.select_dtypes(include=["int","float"])
```

```python
# 24.Fill missing numerical values using the median of each numerical column.
df_med=df[["age","income_usd","credit_score","tenure_years","monthly_visits","satisfaction","churn_flag"]].median()
df[["age","income_usd","credit_score","tenure_years","monthly_visits","satisfaction","churn_flag"]]=df[["age","income_usd","credit_score","tenure_years","monthly_visits","satisfaction","churn_flag"]].fillna(df_med)
df[["age","income_usd","credit_score","tenure_years","monthly_visits","satisfaction","churn_flag"]]
```

```python
# 25.Fill missing categorical values using the mode of each categorical column.
df_mean=df[["customer_id","city","customer_segment","gender","is_active","signup_date","lead_source","notes_text"]].mode()
df[["customer_id","city","customer_segment","gender","is_active","signup_date","lead_source","notes_text"]]=df[["customer_id","city","customer_segment","gender","is_active","signup_date","lead_source","notes_text"]].fillna(df_mean)
df[["customer_id","city","customer_segment","gender","is_active","signup_date","lead_source","notes_text"]]
```

```python
# 26.Clean the signup_date column, which contains multiple date formats and
# invalid dates, and convert it to Pandas datetime.
df["signup_date"]=pd.to_datetime(df["signup_date"],format="mixed",errors="coerce")
df["signup_date"]
```

```python
# 27.Find rows where signup_date could not be converted into a valid date.
df["signup_date"].isna()
```

```python
# 28.Clean the lead_source column by standardizing values such as Website,
# website, WEB, Referral, referral , Google Ads, and google_ads.
df["lead_source"]=df["lead_source"].replace({"website":"Website","WEB":"Website","referral":"Referral","google_ads":"Google Ads","email":"Email"})
df["lead_source"]
```

```python
# 29.Create a final cleaned DataFrame containing:
# • cleaned column names
# • no duplicate IDs
# • standardized categorical values
# • numeric columns converted to numeric types
# • invalid values converted to NaN
# • missing values handled
# • detected outliers handled
df.drop_duplicates(inplace=True)
df.dtypes
```
