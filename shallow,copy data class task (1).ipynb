```python
#1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change
#shallow[2][0] = 99. Print both lists and explain the result.
import copy
original = [1, 2, [3, 4], 5]
shallow=copy.copy(original)
shallow[2][0]=99
print("original list :",original)
print("shallow list :",shallow)
```

```python
#2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.
deep=copy.deepcopy(original)
deep[2][0]=99
print("original list :",original)
print("shallow list :",deep)
```

```python
#5.Attributes• account_holder , balance Methods• deposit()• withdraw()
#Withdraw should happen only if balance is sufficient.
class Bank_acc:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print(f"deposited {amount}")
        print(f" total balance is {self.balance}")

    def withdraw(self,amount):
        if self.balance>=amount:
            self.balance-=amount
            print("withdrawn money",amount)
            print("remaining balance",self.balance)
        else:
            print("CANT WITHDRAW INSUFFICIENT BALANCE")

b=Bank_acc("vaishnavi",2000000)
b.deposit(20000)
b.withdraw(200)
```

```python
#9. Library Attributes• book_name• available_copies
#Method borrow_book()
#Reduce copy only if available.

class Library:
    def __init__(self,book_name,available_copies):
        self.book_name=book_name
        self.available_copies=available_copies

    def borrow_book(self,borrow_copy):
        if self.book_name==self.available_copies:
            self.available_copies-=borrow_copy
            print("the new copies are",borrow_copy)
        else:
            print("no copies available")

L=Library("the silent patient",23)
L.borrow_book(2)
        
```

```python
#10. Product
#Attributes • product_name • quantity
#Method sell(quantity)
#Quantity should never become negative.
class Product:
    def __init__(self,product_name,quantity):
        self.product_name=product_name
        self.quantity=quantity
    def sell(self,A_quantity):
        if self.quantity<=0:
            print("Invalid quantity! it cant ne negative")
        else:
            self.quantity-=A_quantity
            print(f"the quantity of product is {A_quantity}")

P=Product("pencil",34)
P.sell(145)
```

```python
# 17. School Fees
# Attributes
# • student_name
# • total_fee
# • paid_fee
# Method
# # pay_fee(amount)
# # Fee paid cannot exceed total fee.
class Fees:
    def __init__(self,student_name,total_fee,paid_fee):
        self.student_name=student_name
        self.total_fee=total_fee
        self.paid_fee=paid_fee
        
    def pay_fee(self,amount):
        if self.total_fee==self.paid_fee:
            print("fees paid")
        elif amount>=self.total_fee:
            print("INVALID! PAID AMOUNT EXCEED FEES")
        else:
            amount-=self.total_fee
            print("balance fees is",amount)

F=Fees("reansh",20000,10000)
F.pay_fee(8000)
```

```python
# 19. ATM
# Attributes
# • balance
# Methods
# deposit()
# withdraw()
# withdraw()
# should reject
# • negative amount
# • zero
# # • insufficient balance

class ATM:
    def __init__(self,balance):
        self.balance=balance
    def deposit(self,dep):
        self.balance+=dep
        print("deposited amount",self.balance)
    def withdraw(self):
        w_draw=int(input("enter amount"))
        if w_draw<0:
            print("invalid")
        elif w_draw<=0:
            print("invalid")
        elif w_draw>=self.balance:
            print("insufficient amount")
        self.balance=self.balance-w_draw
        print(self.balance)


A=ATM(20000)
A.deposit(200)
A.withdraw()
        
```

```python
# 21. Inventory System
# Attributes
# • product_name
# • stock
# Method
# update_stock(change)
# Positive val become negative.
class Inventory:
    def __init__(self,product_name,stock):
        self.product_name=product_name
        self.stock=stock
    def update_stock(self,change):
        if change>0:
            change-=change
            
        if self.stock+change<0:
            print("stock can never be negative")
        else:
            self.stock+=change
            print(self.stock)
        

I=Inventory("pen",23)
I.update_stock(-3)

```

```python
# 24. Bus Reservation (Hard)
# Attributes
# • bus_name
# • total_seats
# • booked_seats
# Method
# reserve(number)
# Reservation allowed only if
# booked_seats + number
# does not exceed total seats.
# After booking print
# Remaining seats.
class Bus:
    def __init__(self,bus_name,total_seats,booked_seats):
        self.bus_name=bus_name
        self.total_seats=total_seats
        self.booked_seats=booked_seats
    def reserve(self,number):
        number=int(input("enter the number"))
        if self.booked_seats+number<=self.total_seats:
            print("you can reserve the seat")
            rem=self.total_seats-(self.booked_seats+number)
            print("The remaining seats are=",rem)

        else:
            print("INVALID SEATS ARE FULL")
bus=Bus("MH1234",79,50)
bus.reserve(5)
        
```
