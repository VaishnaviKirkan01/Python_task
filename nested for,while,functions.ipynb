parameters types

```python
#1 NO PARAMETERS
def hello():
    return"good morning"

hello()
```

```python
#2 POSITIONAL PARAMETERS
def info(name,age):
    return f"hii my name is {name} and age is {age}"

info("neha",29)
```

```python
#3 DEFAULT PARAMETERS
def hello(name="sumit"):
    return f"hello this is {name}"

hello()
```

```python
#4 KEYWORD ARGUMENT
def info(name,age):
    return f"hii my name is {name} and age is {age}"
info(age=37,name='neha')
```

```python
#take the name,address,city,cource,mentor_name
def info(name,city,address,cource,mentor_name):
    print(f"hii my name is = {name} . \nmy city is = {city}. \nthe addresss is = {address}. \nmy cource is = {cource} \nand the course mentor is = {mentor_name} ")
   
info(name="neha",city="mumbai",address="delhi",cource="data science",mentor_name="sumit sir")
```

```python
#5 MULTI POSITIONAL ARGUMENTS
def summ(*numbers):
    c=sum(numbers)
    return c
summ(1,6,5,4,3,3,9,2,2,2,2,9,8)
```

```python
#6 MULIPLE KEYWORD ARGUM:ENT
def info(**details):
    for key,value in details.item():
        return f" key = {key} : value = {value} "
        

info("name":"neha","city":"mumbai","address":"delhi","cource":"data science","mentor_name":"sumit sir")
```
