```python
# 10.Convert Date_of_Journey into a datetime format.
df["Date_of_Journey"]=pd.to_datetime(df["Date_of_Journey"],errors="coerce")
df["Date_of_Journey"]
```

```python
# 13.Extract the year from Date_of_Journey and create a year column.
df["year"]=df["Date_of_Journey"].dt.year
df
```

```python
# 14.Drop the original Date_of_Journey column after extracting the required
# features.
df.drop("Date_of_Journey",axis=1,inplace=True)
```

```python
# 15.Convert Dep_Time into datetime format and extract the departure hour into
# Dep_Hour.
df["Dep_Time"]=pd.to_datetime(df["Dep_Time"],errors="coerce")
df["Dep_Time"]
```

```python
df["Dep_Hour"]=df["Dep_Time"].dt.hour
df
```

```python
# 16.Extract the departure minute from Dep_Time and create Dep_Minute.
df["Dep_Minute"]=df["Dep_Time"].dt.minute
df
```

```python
# 17.Drop the original Dep_Time column after extracting the required features.
df.drop("Dep_Time",axis=1,inplace=True)
df
```

```python
# 18.Extract only the time portion from Arrival_Time.
#df["Arrival_Time"]=df["Arrival_Time"].str.split().str[0]
df["Arrival_Time"]=pd.to_datetime(df["Arrival_Time"],errors="coerce")
df["Arrival_Time"]
```

```python
# 19.Extract the arrival hour and minute into Arrival_Hour and Arrival_Minute.
df["Arrival_Hour"]=df["Arrival_Time"].dt.hour
df
```

```python
df["Arrival_Minute"]=df["Arrival_Time"].dt.minute
df
```

```python
# 20.Drop the original Arrival_Time column after extracting the required features.
df.drop("Arrival_Time",axis=1,inplace=True)
```

```python
df
```

```python
# 21.Extract the hour component from Duration and create Duration_Hour. Handle
# invalid or missing extracted values appropriately.
df["Duration_Hour"]=df["Duration"].str.split("h").str[0]
```

```python
df["Duration_Hour"]=pd.to_numeric(df["Duration_Hour"],errors="coerce")
```

```python
due_median=df["Duration_Hour"].median()
df["Duration_Hour"]=df["Duration_Hour"].fillna(due_median)
```

```python
df[["Duration_Hour","Duration"]].head()
```

```python
df
```

```python
# 22.Extract the minute component from Duration and create Duration_Minute.
# Handle invalid or missing extracted values appropriately.
df["Duration_Minute"]=df["Duration"].str.split(" ").str[1].str.replace("m"," ")
df["Duration_Minute"]=pd.to_numeric(df["Duration_Minute"],errors="coerce")
min_med=df["Duration_Minute"].median()
df["Duration_Minute"]=df["Duration_Minute"].fillna(min_med)
df
```

```python
# 23.Drop the original Duration column after creating the numerical duration
# features.
df.drop("Duration",axis=1,inplace=True)
```

```python
df
```

```python
# 24.Replace non-stop in Total_Stops with 0 stops.
df.replace({"Total_Stops":{"non-stop":0}},inplace=True)
df
```

```python
# 25.Extract only the numerical value from Total_Stops and convert the column into
# an integer data type. Handle missing or invalid values appropriately.

df["Total_Stops"]=df["Total_Stops"].replace({
    "non-stop":0,"1 stop":1,"2 stops":2,"3 stops":3,"4 stops":4
})
```

```python

df["Total_Stops"]=df["Total_Stops"].fillna(0)
```

```python

df["Total_Stops"]=df["Total_Stops"].astype("int")
```

```python
df["Total_Stops"].head()
```

```python
df["Total_Stops"].unique()
```

```python
# 26.Split the Route column using → as the separator and create Route_1,
# Route_2, Route_3, Route_4, and Route_5.
route=df["Route"].str.split("→",expand=True)
```

```python
df["Route_1"]=route[0].str.strip()
df["Route_2"]=route[1].str.strip()
df["Route_3"]=route[2].str.strip()
df["Route_4"]=route[3].str.strip()
df["Route_5"]=route[4].str.strip()
df["Route_6"]=route[5].str.strip()
```

```python
df
```

```python
# 27.Identify the missing values created in Route_3, Route_4, and Route_5 after
# splitting the route.
df[["Route_3","Route_4","Route_5"]].isnull().sum()
```

```python
df
```

```python
# 28.Replace the missing values in the route columns with an appropriate value
# such as None.
df[["Route_1","Route_2","Route_3","Route_4","Route_5"]]=df[["Route_1","Route_2","Route_3","Route_4","Route_5"]].fillna("None")
```
