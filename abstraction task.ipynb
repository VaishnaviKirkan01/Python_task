```python
# 1. Class Name: Shape
# Problem
# Create an abstract class named Shape.
# Explanation
# A shape can have different formulas for area and perimeter. Therefore, create two abstract
# methods:
# • area()
# • perimeter()
# Create subclasses like Circle and Rectangle and implement both methods.
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area():
        pass
    def perimeter():
        pass

class Circle(Shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        area=3.14*self.r
        print("The area of circle is= ",area)
    def perimeter(self):
        peri=2*3.14*self.r
        print("The circle perimeter is= ",peri)

class Rectangle(Shape):
    def __init__(self,l,w):
        self.length=l
        self.width=w
    def area(self):
        area=self.length*self.width
        print("The rectangle area is= ",area)
    def perimeter(self):
        peri=2*(self.length+self.width)
        print("The rectabgle perimeter= ",peri)

cir=Circle(3)
cir.area()
cir.perimeter()

rect=Rectangle(3,5)
rect.area()
rect.perimeter()
```
